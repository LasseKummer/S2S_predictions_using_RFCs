```python
import pandas as pd
import datetime as dt
```

```python
df_mjo = pd.read_csv('MJO_week3.csv')
df_enso = pd.read_csv('ENSO_week3.csv')
df_nao = pd.read_csv('NAO_week3.csv')
df_arctic = pd.read_csv('arctic_sea_ice_week3.csv')
df_baka = pd.read_csv('baka_sea_ice_week3.csv')
df_u10 = pd.read_csv('u10_60N_week3.csv')

# Choose region:
df_sm1 = pd.read_csv('sm_upper_NE_week3_scaled.csv')
df_sm2 = pd.read_csv('sm_lower_NE_week3_detr.csv')
df_snow = pd.read_csv('snow_NE_week3.csv')
```

```python
df_features = pd.concat([df_mjo, df_enso.iloc[:,1], df_nao.iloc[:,1], df_arctic.iloc[:,1], df_baka.iloc[:,1], df_sm1.iloc[:,1], 
                         df_sm2.iloc[:,1], df_snow.iloc[:,1], df_u10.iloc[:,1]], axis=1)
```

```python
df_features = df_features.rename(columns={'Unnamed: 0':'time'})
df_features['time'] = pd.to_datetime(df_features['time'])
df_features['time'] = df_features['time'] + dt.timedelta(days=18)
df_features = df_features.set_index('time')
df_features
```

```python
# Name appropriately:

df_features.to_csv('features_NE_week3_sens.csv')
```
