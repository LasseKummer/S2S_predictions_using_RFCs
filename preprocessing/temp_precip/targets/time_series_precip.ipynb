```python
var_daily = var_hourly.groupby(var_hourly.valid_time.dt.date).sum()
```

```python
dates = var_daily.date.data
datetimes = np.zeros_like(dates)
for i in range(len(dates)):
    datetimes[i] = datetime.combine(dates[i], datetime.min.time())
```

```python
var_daily.coords["time"] = ("date", datetimes)
var_daily = var_daily.swap_dims({"date": "time"})
var_daily = var_daily.reset_coords('date', drop=True)
```

```python
var_daily = var_daily.sel(time=slice("1940-07", "2025-06"))
month = var_daily.time.dt.month
day = var_daily.time.dt.day
var_daily = var_daily.sel(time=(month==12)|(month==1)|((month==2) & (day<29)))
```

### Spatial weights

```python
ds_oro = xr.load_dataset("target_variables/grib_files/orography_original_res.grib", engine="cfgrib")
```

```python
g = 9.80665
```

```python
ds_oro = ds_oro.reset_coords(['number', 'time', 'step', 'surface', 'valid_time'], drop=True)
```

```python
lat = ds_oro.latitude
land_frac = ds_oro.lsm + ds_oro.cl
mount_filt = xr.where(ds_oro.z/g < 1000, 1, 0)
```

```python
weights = np.cos(np.deg2rad(lat))*land_frac*mount_filt
weights.name = "weights"
```

```python
weights.plot()
```

### Time series of region means

**Northern Europe**: 4°E-42°E, 59°N-72°N  
**Western-Central Europe**: 11°W-19°E, 47°N-59°N  
**Eastern Europe**: 19°E-42°E, 47°N-59°N  
**Southwestern Europe**: 10°W-10°E, 36°N-47°N  
**Southeastern Europe**: 10°E-30°E, 35°N-47°N

```python
# Choose extent of the region:

lon_min = 10
lon_max = 30
lat_max = 47
lat_min = 35

var_daily_regmean = var_daily.sel(longitude=slice(lon_min,lon_max), latitude=slice(lat_max,lat_min)).weighted(weights).mean(dim=['longitude', 'latitude'])
```

```python
var_7dMean_regmean = var_daily_regmean.rolling(time=7, min_periods=7, center=True).mean()
month = var_7dMean_regmean.time.dt.month
day = var_7dMean_regmean.time.dt.day
var_7dMean_regmean = var_7dMean_regmean.sel(time=~( ((month==12)&(day<4)) | ((month==2)&(day>25)) ))
```

```python
var_7dMean_regmean.plot()
```

```python
var_seasonmean_regmean = var_7dMean_regmean.coarsen(time=84).construct(time=("year", "date")).mean(dim='date')
plt.plot(var_seasonmean_regmean.year, var_seasonmean_regmean)
```

```python
fit_par = np.polyfit(var_seasonmean_regmean[10:70].year, var_seasonmean_regmean[10:70], 1)
var_seasonmean_regmean_gws = np.polyval(fit_par, var_seasonmean_regmean.year)

plt.plot(var_seasonmean_regmean.year, var_seasonmean_regmean)
plt.plot(var_seasonmean_regmean.year, var_seasonmean_regmean_gws)
```

```python
var_7dMean_regmean_detr = var_7dMean_regmean.coarsen(time=84).construct(time=("year", "date")) - var_seasonmean_regmean_gws[:, np.newaxis]
var_7dMean_regmean_detr = var_7dMean_regmean_detr.stack(day=("year", "date")).swap_dims({"day": "time"}).reset_coords(('day','year','date'), drop=True)
var_7dMean_regmean_detr.sel(time=slice("1950-07", "2010-06")).plot()
```

```python
plt.hist(var_7dMean_regmean.sel(time=slice("1950-07", "2010-06")), bins='auto', density=True)
plt.show()
```

```python
var_7dMean_regmean_train = var_7dMean_regmean.sel(time=slice("1950-07", "2010-06"))

bc_transformer = PowerTransformer(method="box-cox")
transformed_values = bc_transformer.fit(var_7dMean_regmean_train.values.reshape(-1, 1)).transform(var_7dMean_regmean.values.reshape(-1, 1))
```

```python
transvar_7dMean_regmean = var_7dMean_regmean.copy()
transvar_7dMean_regmean[:] = transformed_values[:,0]
```

```python
transvar_7dMean_regmean.plot()
```

```python
transvar_seasonmean_regmean = transvar_7dMean_regmean.coarsen(time=84).construct(time=("year", "date")).mean(dim='date')
plt.plot(transvar_seasonmean_regmean.year, transvar_seasonmean_regmean)
```

```python
fit_par = np.polyfit(transvar_seasonmean_regmean[10:70].year, transvar_seasonmean_regmean[10:70], 1)
transvar_seasonmean_regmean_gws = np.polyval(fit_par, transvar_seasonmean_regmean.year)

plt.plot(transvar_seasonmean_regmean.year, transvar_seasonmean_regmean)
plt.plot(transvar_seasonmean_regmean.year, transvar_seasonmean_regmean_gws)
```

```python
transvar_7dMean_regmean_detr = transvar_7dMean_regmean.coarsen(time=84).construct(time=("year", "date")) - transvar_seasonmean_regmean_gws[:, np.newaxis]
transvar_7dMean_regmean_detr = transvar_7dMean_regmean_detr.stack(day=("year", "date")).swap_dims({"day": "time"}).reset_coords(('day','year','date'), drop=True)
transvar_7dMean_regmean_detr.sel(time=slice("1950-07", "2010-06")).plot()
```

```python
transvar_7dMean_regmean_detr_train = transvar_7dMean_regmean_detr.sel(time=slice("1950-07", "2010-06"))
transvar_7dMean_regmean_detr_monthlyanom = transvar_7dMean_regmean_detr.groupby('time.month') - transvar_7dMean_regmean_detr_train.groupby('time.month').mean()
transvar_7dMean_regmean_detr_monthlyanom.plot()
```
