```python
import numpy as np
import pandas as pd
```

```python
# Choose correct data files:

df_precip = pd.read_csv('target_precip_EE_lin_moanom.csv')
```

```python
df_precip['time'] = pd.to_datetime(df_precip['time'])
```

```python
df_precip_train = df_precip[df_precip['time'] < "2010-07"]
```

```python
lower_tercile_precip = df_precip_train.quantile([1/3])['tp'].values[0]
print(lower_tercile_precip)
upper_tercile_precip = df_precip_train.quantile([2/3])['tp'].values[0]
print(upper_tercile_precip)
```

```python
df_precip['tercile_category'] = 1
df_precip['tercile_category'] = df_precip['tercile_category'].where(df_precip['tp'] > lower_tercile_precip, 0)
df_precip['tercile_category'] = df_precip['tercile_category'].where(df_precip['tp'] < upper_tercile_precip, 2)
```

```python
df_precip
```

```python
df_precip = df_precip.drop(['tp'], axis=1).set_index('time')
```

```python
print(len(df_precip[(df_precip.index < "2010-07") & (df_precip['tercile_category'] == 0)])/5040)
print(len(df_precip[(df_precip.index < "2010-07") & (df_precip['tercile_category'] == 1)])/5040)
print(len(df_precip[(df_precip.index < "2010-07") & (df_precip['tercile_category'] == 2)])/5040)
```
