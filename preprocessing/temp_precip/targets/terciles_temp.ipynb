```python
import numpy as np
import pandas as pd
```

```python
# Choose correct data files:

df_temp = pd.read_csv('target_temp_SWE_quadr_moanom.csv')
#df_precip = pd.read_csv('ground_truth_time_series/southwestern_precipitation.csv')
```

```python
df_temp['time'] = pd.to_datetime(df_temp['time'])
#df_precip['time'] = pd.to_datetime(df_precip['time'])
```

```python
df_temp_train = df_temp[df_temp['time'] < "2010-07"]
#df_precip_train = df_precip[df_precip['time'] < "2010-07"]
```

```python
lower_tercile_temp = df_temp_train.quantile([1/3])['t2m'].values[0]
print(lower_tercile_temp)
upper_tercile_temp = df_temp_train.quantile([2/3])['t2m'].values[0]
print(upper_tercile_temp)
```

```python
df_temp['tercile_category'] = 1
df_temp['tercile_category'] = df_temp['tercile_category'].where(df_temp['t2m'] > lower_tercile_temp, 0)
df_temp['tercile_category'] = df_temp['tercile_category'].where(df_temp['t2m'] < upper_tercile_temp, 2)
```

```python
df_temp
```

```python
df_temp = df_temp.drop(['t2m'], axis=1).set_index('time')
#df_precip = df_precip.drop('rr', axis=1).set_index('time')
```

```python
# Name data files appropriately:

df_temp.to_csv('southeastern_temperature_terc_cat_new.csv')
#df_precip.to_csv('tercile_category_time_series/southwestern_precipitation_terc_cat.csv')
```
