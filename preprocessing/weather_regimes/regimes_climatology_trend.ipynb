```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('weather_regimes_target.csv')
```

```python
reshaped_data = df['NAE_regime'].values.reshape((5,84*15))
```

```python
plt.plot(range(5), np.count_nonzero(reshaped_data == 0, axis=1)/1260)
plt.plot(range(5), np.count_nonzero(reshaped_data == 2, axis=1)/1260)
plt.ylim(bottom=0)
```

```python
plt.plot(range(5), np.count_nonzero(reshaped_data == 1, axis=1)/1260)
plt.plot(range(5), np.count_nonzero(reshaped_data == 3, axis=1)/1260)
plt.ylim(bottom=0)
```

```python
reshaped_data = df['NAE_regime'].values.reshape((75,84))
```

```python
plt.plot(range(84), np.count_nonzero(reshaped_data[:60] == 0, axis=0)/60)
plt.plot(range(84), np.count_nonzero(reshaped_data[:60] == 2, axis=0)/60)
plt.ylim(bottom=0)
```

```python
plt.plot(range(84), np.count_nonzero(reshaped_data[60:] == 0, axis=0)/15)
plt.plot(range(84), np.count_nonzero(reshaped_data[60:] == 2, axis=0)/15)
plt.ylim(bottom=0)
```

```python
plt.plot(range(84), np.count_nonzero(reshaped_data[:60] == 1, axis=0)/60)
plt.plot(range(84), np.count_nonzero(reshaped_data[:60] == 3, axis=0)/60)
plt.ylim(bottom=0)
```

```python
plt.plot(range(84), np.count_nonzero(reshaped_data[60:] == 1, axis=0)/15)
plt.plot(range(84), np.count_nonzero(reshaped_data[60:] == 3, axis=0)/15)
plt.ylim(bottom=0)
```
