```python
import pandas as pd
import datetime as dt
```

```python
df_mjo = pd.read_csv('../Part1_analysis/MJO_week5.csv')
df_enso = pd.read_csv('../Part1_analysis/ENSO_week5.csv')
df_nao = pd.read_csv('../Part1_analysis/NAO_week5.csv')
df_arctic = pd.read_csv('../Part1_analysis/arctic_sea_ice_week5.csv')
df_baka = pd.read_csv('../Part1_analysis/baka_sea_ice_week5.csv')
df_u10 = pd.read_csv('../Part1_analysis/u10_60N_week5.csv')
df_st100 = pd.read_csv('st100_index_week5.csv')
df_snow = pd.read_csv('Siberian_snow_week5.csv')
```

```python
df_features = pd.concat([df_mjo, df_enso.iloc[:,1], df_nao.iloc[:,1], df_arctic.iloc[:,1], df_baka.iloc[:,1], df_u10.iloc[:,1], 
                        df_st100.iloc[:,1], df_snow.iloc[:,1]], axis=1)
```

```python
df_features = df_features.rename(columns={'Unnamed: 0':'time', 'snow_depth':'siberian_snow'})
df_features['time'] = pd.to_datetime(df_features['time'])
df_features['time'] = df_features['time'] + dt.timedelta(days=32)
df_features = df_features.set_index('time')
df_features
```

```python
# Name appropriately:

df_features.to_csv('features_regimes_week5.csv')
```
