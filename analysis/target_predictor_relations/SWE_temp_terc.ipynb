```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df_temp = pd.read_csv('../Part1_analysis/target_temp_SWE_quadr_moanom.csv')
df_feature = pd.read_csv('../Part1_analysis/arctic_sea_ice_week3.csv')
df_temp_terc = pd.read_csv('../Part1_analysis/southwestern_temperature_terc_cat_new.csv')
```

```python
plt.plot(df_feature['sea_ice_arctic'][:5040], df_temp['t2m'][:5040], 'o', markersize=1, label='1950-2010')
plt.plot(df_feature['sea_ice_arctic'][5040:], df_temp['t2m'][5040:], 'o', markersize=1, label='2010-2025')
plt.xlabel('week 0 Arctic sea ice concentration')
plt.ylabel('week 3 temperature at 2m [K]')
plt.title('Southwestern Europe, DJF')
plt.legend()
plt.grid()
plt.savefig('SWE_temp_arctic_scatter.png', bbox_inches='tight')
```

```python
low_temp_train = df_feature[:5040][df_temp_terc[:5040]['tercile_category']==0]['sea_ice_arctic']
medium_temp_train = df_feature[:5040][df_temp_terc[:5040]['tercile_category']==1]['sea_ice_arctic']
high_temp_train = df_feature[:5040][df_temp_terc[:5040]['tercile_category']==2]['sea_ice_arctic']

low_temp_test = df_feature[5040:][df_temp_terc[5040:]['tercile_category']==0]['sea_ice_arctic']
medium_temp_test = df_feature[5040:][df_temp_terc[5040:]['tercile_category']==1]['sea_ice_arctic']
high_temp_test = df_feature[5040:][df_temp_terc[5040:]['tercile_category']==2]['sea_ice_arctic']
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,5))

ax[0].boxplot([low_temp_train, medium_temp_train, high_temp_train], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[0].set_xlabel('week 0 Arctic sea ice concentration',fontsize=12)
ax[0].set_ylabel('week 3 temperature class',fontsize=12)
ax[0].set_title('Southwestern Europe, DJF 1950-2010',fontsize=14)
ax[0].set_xlim((-0.035,0.052))
ax[0].tick_params(labelsize=11)

ax[1].boxplot([low_temp_test, medium_temp_test, high_temp_test], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[1].set_xlabel('week 0 Arctic sea ice concentration',fontsize=12)
ax[1].set_title('Southwestern Europe, DJF 2010-2025',fontsize=14)
ax[1].set_xlim((-0.035,0.052))
ax[1].tick_params(labelsize=11)

plt.savefig('SWE_temp_arctic_box.pdf', bbox_inches='tight')
```
