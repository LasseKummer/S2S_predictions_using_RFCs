```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df_temp_NE = pd.read_csv('../Part1_analysis/target_temp_NE_quadr_moanom.csv')
df_sm_upper_NE = pd.read_csv('../Part1_analysis/sm_upper_NE_week5_detr.csv')
df_temp_terc_NE = pd.read_csv('../Part1_analysis/northern_temperature_terc_cat_new.csv')
```

```python
plt.plot(df_sm_upper_NE['soil_moist_upper'][:5040], df_temp_NE['t2m'][:5040], 'o', markersize=1, label='1950-2010')
plt.plot(df_sm_upper_NE['soil_moist_upper'][5040:], df_temp_NE['t2m'][5040:], 'o', markersize=1, label='2010-2025')
plt.xlabel('week 0 soil moisture fraction, layers 1+2')
plt.ylabel('week 3 temperature at 2m [K]')
plt.title('Northern Europe, DJF')
plt.legend()
plt.grid()
#plt.savefig('NE_temp_sm_scatter.png', bbox_inches='tight')
```

```python
low_temp_train = df_sm_upper_NE[:5040][df_temp_terc_NE[:5040]['tercile_category']==0]['soil_moist_upper']
medium_temp_train = df_sm_upper_NE[:5040][df_temp_terc_NE[:5040]['tercile_category']==1]['soil_moist_upper']
high_temp_train = df_sm_upper_NE[:5040][df_temp_terc_NE[:5040]['tercile_category']==2]['soil_moist_upper']

low_temp_test = df_sm_upper_NE[5040:][df_temp_terc_NE[5040:]['tercile_category']==0]['soil_moist_upper']
medium_temp_test = df_sm_upper_NE[5040:][df_temp_terc_NE[5040:]['tercile_category']==1]['soil_moist_upper']
high_temp_test = df_sm_upper_NE[5040:][df_temp_terc_NE[5040:]['tercile_category']==2]['soil_moist_upper']
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,5))

ax[0].boxplot([low_temp_train, medium_temp_train, high_temp_train], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[0].set_xlabel('week 0 soil moisture fraction, layers 1+2')
ax[0].set_ylabel('week 3 temperature class')
ax[0].set_title('Northern Europe, DJF 1950-2010')
ax[0].set_xlim((-0.03,0.045))

ax[1].boxplot([low_temp_test, medium_temp_test, high_temp_test], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[1].set_xlabel('week 0 soil moisture fraction, layers 1+2')
ax[1].set_title('Northern Europe, DJF 2010-2025')
ax[1].set_xlim((-0.03,0.045))

#plt.savefig('NE_temp_sm_box.png', bbox_inches='tight')
```
