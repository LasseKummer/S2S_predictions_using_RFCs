```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df_target = pd.read_csv('../weather_regimes_analysis/weather_regimes_target.csv')
```

```python
df_target
```

```python
df_nao = pd.read_csv('../Part1_analysis/NAO_week3.csv')
```

```python
NAOneg_train = df_nao[:5040][df_target[:5040]['NAE_regime']==0]['nao_index']
AtlRidge_train = df_nao[:5040][df_target[:5040]['NAE_regime']==1]['nao_index']
NAOpos_train = df_nao[:5040][df_target[:5040]['NAE_regime']==2]['nao_index']
ScandBlock_train = df_nao[:5040][df_target[:5040]['NAE_regime']==3]['nao_index']

NAOneg_test = df_nao[5040:][df_target[5040:]['NAE_regime']==0]['nao_index']
AtlRidge_test = df_nao[5040:][df_target[5040:]['NAE_regime']==1]['nao_index']
NAOpos_test = df_nao[5040:][df_target[5040:]['NAE_regime']==2]['nao_index']
ScandBlock_test = df_nao[5040:][df_target[5040:]['NAE_regime']==3]['nao_index']
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,5))

ax[0].boxplot([NAOneg_train, AtlRidge_train, NAOpos_train, NAOneg_train], orientation='horizontal', whis=[5,95], sym="", 
               tick_labels=['NAO negative', 'Atlantic Ridge', 'NAO positive', 'Scandinavian Blocking'])
ax[0].set_xlabel('week 0 NAO index',fontsize=12)
ax[0].set_ylabel('week 3 weather regime',fontsize=12)
ax[0].set_title('DJF 1950-2010',fontsize=14)
ax[0].set_xlim((-1.6,1.8))
ax[0].tick_params(labelsize=11)

ax[1].boxplot([NAOneg_test, AtlRidge_test, NAOpos_test, NAOneg_test], orientation='horizontal', whis=[5,95], sym="", 
               tick_labels=['','','',''])
ax[1].set_xlabel('week 0 NAO index',fontsize=12)
ax[1].set_title('DJF 2010-2025',fontsize=14)
ax[1].set_xlim((-1.6,1.8))
ax[1].tick_params(labelsize=11)

plt.savefig('regimes_nao_box.pdf', bbox_inches='tight')
```
