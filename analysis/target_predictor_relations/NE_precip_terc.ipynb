```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df_precip_NE = pd.read_csv('../Part1_analysis/target_precip_NE_lin_moanom.csv')
df_sm_upper_NE = pd.read_csv('../Part1_analysis/sm_upper_NE_week3_detr.csv')
df_precip_terc_NE = pd.read_csv('../Part1_analysis/northern_precipitation_terc_cat_trans.csv')
```

```python
plt.plot(df_sm_upper_NE['soil_moist_upper'][:5040], df_precip_NE['tp'][:5040], 'o', markersize=1, label='1950-2010')
plt.plot(df_sm_upper_NE['soil_moist_upper'][5040:], df_precip_NE['tp'][5040:], 'o', markersize=1, label='2010-2025')
plt.xlabel('week 0 soil moisture fraction, layers 1+2')
plt.ylabel('week 3 precipitation (transformed)')
plt.title('Northern Europe, DJF')
plt.legend()
plt.grid()
plt.savefig('NE_precip_sm_scatter.png', bbox_inches='tight')
```

```python
low_precip_train = df_sm_upper_NE[:5040][df_precip_terc_NE[:5040]['tercile_category']==0]['soil_moist_upper']
medium_precip_train = df_sm_upper_NE[:5040][df_precip_terc_NE[:5040]['tercile_category']==1]['soil_moist_upper']
high_precip_train = df_sm_upper_NE[:5040][df_precip_terc_NE[:5040]['tercile_category']==2]['soil_moist_upper']

low_precip_test = df_sm_upper_NE[5040:][df_precip_terc_NE[5040:]['tercile_category']==0]['soil_moist_upper']
medium_precip_test = df_sm_upper_NE[5040:][df_precip_terc_NE[5040:]['tercile_category']==1]['soil_moist_upper']
high_precip_test = df_sm_upper_NE[5040:][df_precip_terc_NE[5040:]['tercile_category']==2]['soil_moist_upper']
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,5))

ax[0].boxplot([low_precip_train, medium_precip_train, high_precip_train], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[0].set_xlabel('week 0 soil moisture fraction, IFS layers 1+2',fontsize=12)
ax[0].set_ylabel('week 3 precipitation class',fontsize=12)
ax[0].set_title('Northern Europe, DJF 1950-2010',fontsize=14)
ax[0].set_xlim((-0.035,0.06))
ax[0].tick_params(labelsize=11)

ax[1].boxplot([low_precip_test, medium_precip_test, high_precip_test], orientation='horizontal', whis=[5,95], sym="", tick_labels=['low', 'medium', 'high'])
ax[1].set_xlabel('week 0 soil moisture fraction, IFS layers 1+2',fontsize=12)
ax[1].set_title('Northern Europe, DJF 2010-2025',fontsize=14)
ax[1].set_xlim((-0.035,0.06))
ax[1].tick_params(labelsize=11)

plt.savefig('NE_precip_sm_box.pdf', bbox_inches='tight')
```
